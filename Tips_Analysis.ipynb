# Tips

In this task I look at the bills of a restaurant and see what changes the tip.

Columns I will use:

| Column | Meaning |
|---|---|
| total_bill | bill in dollars |
| tip | tip in dollars |
| sex | sex of the payer |
| smoker | smoker or not |
| day | day of the week |
| time | lunch or dinner |
| size | number of people |

## Libraries

- pandas: tables and csv files
- numpy: arrays
- plotly express: plots
- seaborn: only to download the data

```python
import pandas as pd
import numpy as np
import plotly.express as px
import seaborn as sns
```

## Loading the data

First I download the data, save it as csv, and read it again with pandas.

```python
sns.load_dataset("tips").to_csv("tips.csv", index=False)

df = pd.read_csv("tips.csv")
df.head()
```

## DataFrame

```python
df.shape
```

```python
df.info()
```

```python
df.isna().sum()
```

I add a new column for the tip percent.

```python
df["tip_percent"] = df["tip"] / df["total_bill"] * 100
df.head()
```

```python
df.sort_values("tip_percent", ascending=False).head()
```

```python
df["day"].value_counts()
```

```python
df.groupby("day")["total_bill"].mean()
```

```python
df.groupby("time")["tip"].mean()
```

```python
df[df["size"] >= 5]
```

## ndarray

```python
bills = df["total_bill"].to_numpy()

print(bills.shape)
print(bills.ndim)
print(bills.mean())
```

```python
expensive = bills[bills > 30]
print(len(expensive))
```

```python
money = df[["total_bill", "tip"]].to_numpy()
print(money.shape)
print(money.sum(axis=0))
```

## Numeric

```python
df[["total_bill", "tip", "size", "tip_percent"]].describe()
```

```python
bill = df["total_bill"]

print("mean:", bill.mean())
print("median:", bill.median())
print("variance:", bill.var())
print("std:", bill.std())
print("skew:", bill.skew())
print("kurtosis:", bill.kurt())
```

```python
df[["total_bill", "tip", "size", "tip_percent"]].corr()
```

## Plots

```python
fig = px.histogram(df, x="total_bill", nbins=20, title="Total bill")
fig.show()
```

```python
fig = px.scatter(df, x="total_bill", y="tip", color="time", title="Tip vs total bill")
fig.show()
```

```python
avg = df.groupby("day")["total_bill"].mean().reset_index()

fig = px.bar(avg, x="day", y="total_bill",
             category_orders={"day": ["Thur", "Fri", "Sat", "Sun"]},
             title="Average bill per day")
fig.show()
```

```python
fig = px.box(df, x="day", y="tip_percent",
             category_orders={"day": ["Thur", "Fri", "Sat", "Sun"]},
             title="Tip percent per day")
fig.show()
```

## Conclusion

- A bigger bill gives a bigger tip.
- The tip percent is almost the same in all days.
- A few tables pay a lot, so the bills are not balanced.
