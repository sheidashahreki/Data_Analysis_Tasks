# Tips

In this task I look at the bills of a restaurant and see what changes the tip.

Columns I will use:

| Column | Meaning |
|---|---|
| total_bill | bill in dollars |
| tip | tip in dollars |
| sex | sex of the payer |
| smoker | smoker or not |
| day | day of the week |
| time | lunch or dinner |
| size | number of people |

## Libraries

- pandas: tables and csv files
- numpy: arrays
- matplotlib: plots
- seaborn: only to download the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading the data

First I download the data, save it as csv, and read it again with pandas.

```python
sns.load_dataset("tips").to_csv("tips.csv", index=False)

df = pd.read_csv("tips.csv")
df.head()
```

## DataFrame

```python
df.shape
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df["tip_percent"] = df["tip"] / df["total_bill"] * 100
df.head()
```

```python
df.sort_values("tip_percent", ascending=False).head()
```

```python
df["day"].value_counts()
```

Average bill for each day.

```python
days = ["Thur", "Fri", "Sat", "Sun"]
avg_bill = []

for d in days:
    day_data = df[df["day"] == d]
    avg_bill.append(day_data["total_bill"].mean())
    print(d, day_data["total_bill"].mean())
```

```python
lunch = df[df["time"] == "Lunch"]
dinner = df[df["time"] == "Dinner"]

print("lunch tip:", lunch["tip"].mean())
print("dinner tip:", dinner["tip"].mean())
```

```python
df[df["size"] >= 5]
```

## ndarray

```python
bills = df["total_bill"].to_numpy()

print(bills.shape)
print(bills.ndim)
print(bills.mean())
```

```python
expensive = bills[bills > 30]
print(len(expensive))
```

```python
money = df[["total_bill", "tip"]].to_numpy()
print(money.shape)
print(money.sum(axis=0))
```

## Numeric

```python
df[["total_bill", "tip", "size", "tip_percent"]].describe()
```

```python
bill = df["total_bill"]

print("mean:", bill.mean())
print("median:", bill.median())
print("variance:", bill.var())
print("std:", bill.std())
print("skew:", bill.skew())
print("kurtosis:", bill.kurt())
```

```python
df[["total_bill", "tip", "size", "tip_percent"]].corr()
```

## Plots

```python
plt.hist(df["total_bill"], bins=20)
plt.title("Total bill")
plt.xlabel("Bill ($)")
plt.ylabel("Number of bills")
plt.show()
```

```python
plt.scatter(df["total_bill"], df["tip"])
plt.title("Tip vs total bill")
plt.xlabel("Total bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

```python
plt.bar(days, avg_bill)
plt.title("Average bill per day")
plt.ylabel("Bill ($)")
plt.show()
```

```python
plt.hist(df["tip_percent"], bins=20)
plt.title("Tip percent")
plt.xlabel("Tip percent")
plt.ylabel("Number of bills")
plt.show()
```

## Conclusion

- A bigger bill gives a bigger tip.
- The tip percent is almost the same for all bills.
- A few tables pay a lot, so the bills are not balanced.
