**Columns**

| Column | Meaning |
|---|---|
| `name` | City name |
| `longitude` | Longitude in degrees (east) |
| `latitude` | Latitude in degrees (north) |
| `population` | Population (2016) |

## 1. Libraries Used

| Library |'
|---|---|
| `pandas` |
| `matplotlib`
| `jupyter`



```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Data Extraction


```python
data = {
    "name": ["Tehran", "Mashhad", "Isfahan", "Karaj", "Shiraz",
             "Tabriz", "Qom", "Ahvaz", "Kermanshah", "Urmia"],
    "longitude": [51.3890, 59.6168, 51.6680, 50.9391, 52.5836,
                  46.2738, 50.8759, 48.6706, 47.0650, 45.0761],
    "latitude": [35.6892, 36.2605, 32.6546, 35.8400, 29.5918,
                 38.0962, 34.6399, 31.3183, 34.3142, 37.5527],
    "population": [8693706, 3001184, 1961260, 1592492, 1565572,
                   1558693, 1201158, 1184788, 946651, 736224],
}

df = pd.DataFrame(data)
df
```

## 3.CSV File



```python
df.to_csv("iran_cities.csv", index=False)

cities = pd.read_csv("iran_cities.csv")
cities
```

## 5.Analysis

```python
print("Total population:", cities["population"].sum())
print("Average population:", int(cities["population"].mean()))
print("Largest city:", cities.loc[cities["population"].idxmax(), "name"])
print("Smallest city:", cities.loc[cities["population"].idxmin(), "name"])
```

## 6. Visualization

```python
plt.figure(figsize=(10, 5))
plt.bar(cities["name"], cities["population"], color="steelblue")
plt.title("Population of Iran's 10 Largest Cities")
plt.xlabel("City")
plt.ylabel("Population")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
